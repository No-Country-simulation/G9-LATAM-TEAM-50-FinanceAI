# **Inferencia del Modelo**

---



Importar librerías

```python
import re
import string
import joblib
```

Cargar modelo y objetos necesarios

```python
modelo = joblib.load("modelo_linear_svc.pkl")

vectorizer = joblib.load("tfidf_vectorizer.pkl")

encoder = joblib.load("label_encoder.pkl")
```

Función de limpieza

```python
def limpiar_texto(texto):
    texto = str(texto).lower()
    texto = re.sub(r"\s+", " ", texto)
    texto = texto.translate(
        str.maketrans("", "", string.punctuation)
    )
    return texto.strip()
```

Función de predicción

```python
def clasificar_transaccion(descripcion):
    """
    Clasifica una transacción utilizando el modelo entrenado.

    Parámetros
    ----------
    descripcion : str
        Descripción de la transacción.

    Retorna
    -------
    str
        Categoría predicha.
    """

    # Limpiar texto
    texto_limpio = limpiar_texto(descripcion)

    # Vectorizar
    texto_vectorizado = vectorizer.transform([texto_limpio])

    # Predicción
    categoria_id = modelo.predict(texto_vectorizado)

    # Convertir a texto
    categoria = encoder.inverse_transform(categoria_id)

    return categoria[0]
```

Probar el modelo

```python
clasificar_transaccion("farmacias similares")
```

```python
ejemplos = [
    "uber viaje aeropuerto",
    "spotify premium",
    "netflix",
    "farmacias similares",
    "oxxo centro",
    "gasolina pemex",
    "colegiatura universidad",
    "amazon compra",
    "renta departamento",
    "telmex internet"
]
for descripcion in ejemplos:
    categoria = clasificar_transaccion(descripcion)

    print(f"{descripcion:30} → {categoria}")
```
