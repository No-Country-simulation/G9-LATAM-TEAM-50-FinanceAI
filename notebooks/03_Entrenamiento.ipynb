# **Entrenamiento de Modelo**


---



Importar librerías

```python
import pandas as pd
import joblib

from scipy import sparse

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

Cargar datos preparados

```python
X_train = sparse.load_npz("X_train_tfidf.npz")
X_test = sparse.load_npz("X_test_tfidf.npz")
```

```python
y_train = pd.read_csv("y_train.csv").squeeze()
y_test = pd.read_csv("y_test.csv").squeeze()
```

```python
encoder = joblib.load("label_encoder.pkl")
```

Verificar que todo cargó correctamente

```python
print(X_train.shape)
print(X_test.shape)

print(y_train.shape)
print(y_test.shape)
```

Creamos el modelo

```python
modelo = LogisticRegression(
    random_state=42,
    max_iter=1000
)
```

**Entrenamiento**

```python
modelo.fit(X_train, y_train)
```

Hacer predicciones

```python
y_pred = modelo.predict(X_test)
```

Medir Presición (Accuracy)

```python
accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy:.4f}")
```

**Reporte de clasificación**

```python
print(
    classification_report(
        y_test,
        y_pred,
        target_names=encoder.classes_
    )
)
```

Guardamos modelo

```python
joblib.dump(modelo, "modelo_logistic_regression.pkl")
```

## Entrenamiento de LinearSVC

Importamos librerias

```python
from sklearn.svm import LinearSVC
```

Creamos el modelo

```python
modelo_svc = LinearSVC(
    random_state=42
)
```

Entrenamos el modelo

```python
modelo_svc.fit(X_train, y_train)
```

Predicciones del modelo

```python
y_pred_svc = modelo_svc.predict(X_test)
```

Presición (Accuracy)

```python
from sklearn.metrics import accuracy_score

accuracy_svc = accuracy_score(y_test, y_pred_svc)

print(f"Accuracy LinearSVC: {accuracy_svc:.4f}")
```

Reporte de clasificación

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        y_test,
        y_pred_svc,
        target_names=encoder.classes_
    )
)
```

**Tabla comparativa entre ambos modelos**

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

resultados = pd.DataFrame({
    "Modelo": [
        "Regresión Logística",
        "LinearSVC"
    ],
    "Accuracy": [
        accuracy_score(y_test, y_pred),
        accuracy_score(y_test, y_pred_svc)
    ],
    "Precision Macro": [
        precision_score(y_test, y_pred, average="macro"),
        precision_score(y_test, y_pred_svc, average="macro")
    ],
    "Recall Macro": [
        recall_score(y_test, y_pred, average="macro"),
        recall_score(y_test, y_pred_svc, average="macro")
    ],
    "F1 Macro": [
        f1_score(y_test, y_pred, average="macro"),
        f1_score(y_test, y_pred_svc, average="macro")
    ]
})

resultados = resultados.round(4)

resultados
```

Guardamos modelo SVC

```python
joblib.dump(modelo_svc, "modelo_linear_svc.pkl")
```
